```python
columns_to_drop = [
    "index",
    "cartodb_id",
    "the_geom",
    "the_geom_webmercator"
]

df_clean = df.drop(columns=columns_to_drop, errors="ignore")

print("Shape after removing unnecessary columns:", df_clean.shape)
```

```python
df_clean = df_clean.rename(columns={
    "cri_rank": "CRI_Rank",
    "cri_score": "CRI_Score",
    "fatalities_per_100k_rank": "Fatalities_Per_100K_Rank",
    "fatalities_per_100k_total": "Fatalities_Per_100K",
    "fatalities_rank": "Fatalities_Rank",
    "fatalities_total": "Total_Fatalities",
    "losses_per_gdp__rank": "Losses_Per_GDP_Rank",
    "losses_per_gdp__total": "Losses_Per_GDP_Pct",
    "losses_usdm_ppp_rank": "Economic_Loss_Rank",
    "losses_usdm_ppp_total": "Economic_Loss_USD_Mn_PPP",
    "rw_country_code": "Country_Code",
    "rw_country_name": "RW_Country_Name"
})

df_clean.head()
```

```python
text_columns = [
    "country",
    "Country_Code",
    "RW_Country_Name"
]

for col in text_columns:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].astype("string").str.strip()
```

```python
print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])

print("\nColumns:")
print(df_clean.columns.tolist())
```

```python
df_clean[
    (df_clean["CRI_Score"] < 0) |
    (df_clean["Total_Fatalities"] < 0) |
    (df_clean["Fatalities_Per_100K"] < 0) |
    (df_clean["Economic_Loss_USD_Mn_PPP"] < 0)
]
```

```python
df_clean["country"] = df_clean["country"].fillna(
    df_clean["RW_Country_Name"]
)
```

```python
df_clean["Country_Code"] = df_clean["Country_Code"].replace(
    ["", " ", "nan", "None"],
    pd.NA
)
```

```python
df_clean = df_clean.drop(columns=["RW_Country_Name"])
```

```python
numeric_columns = [
    "CRI_Rank",
    "CRI_Score",
    "Fatalities_Per_100K_Rank",
    "Fatalities_Per_100K",
    "Fatalities_Rank",
    "Total_Fatalities",
    "Losses_Per_GDP_Rank",
    "Losses_Per_GDP_Pct",
    "Economic_Loss_Rank",
    "Economic_Loss_USD_Mn_PPP"
]

for col in numeric_columns:
    df_clean[col] = pd.to_numeric(
        df_clean[col],
        errors="coerce"
    )
```

```python
print("Final shape:", df_clean.shape)

print("\nMissing values:")
print(df_clean.isnull().sum())

print("\nDuplicate rows:")
print(df_clean.duplicated().sum())

print("\nData types:")
print(df_clean.dtypes)
```

```python
output_path = r"E:\climate_risk_data_analysis\climate_risk_cleaned.csv"

df_clean.to_csv(
    output_path,
    index=False
)

print("Cleaned dataset saved successfully!")
print(output_path)
```
